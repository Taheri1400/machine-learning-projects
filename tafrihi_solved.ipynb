```python
train_data=train_data.iloc[:-1]
```

```python
drop_cols = ['HashPassportNumber_p', 'HashEmail', 'BuyerMobile', 'NationalCode', 'TicketID', 'BillID', 'UserID']
train_data.drop(columns=drop_cols, inplace=True)
test_data.drop(columns=drop_cols, inplace=True)
```

```python
for col in ['Created', 'CancelTime', 'DepartureTime']:
    train_data[col] = pd.to_datetime(train_data[col], errors='coerce')
    test_data[col] = pd.to_datetime(test_data[col], errors='coerce')
```

```python
train_data['hour_created'] = train_data['Created'].dt.hour
test_data['hour_created'] = test_data['Created'].dt.hour
```

```python
train_data['dayofweek_created'] = train_data['Created'].dt.dayofweek
test_data['dayofweek_created'] = test_data['Created'].dt.dayofweek
```

```python
train_data['time_to_departure'] = (train_data['DepartureTime'] - train_data['Created']).dt.total_seconds() / 3600
test_data['time_to_departure'] = (test_data['DepartureTime'] - test_data['Created']).dt.total_seconds() / 3600
```

```python
train_data.fillna({'CouponDiscount': 0, 'VehicleClass': False, 'Male': False, 'Cancel': 0}, inplace=True)
test_data.fillna({'CouponDiscount': 0, 'VehicleClass': False, 'Male': False, 'Cancel': 0}, inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
train_data['TripReason'] = le.fit_transform(train_data['TripReason'])  # Work=1, Int=0
```

```python
cat_cols = ['From', 'To', 'VehicleType', 'Vehicle']
train_data[cat_cols] = train_data[cat_cols].fillna('missing')
test_data[cat_cols] = test_data[cat_cols].fillna('missing')

train_data = pd.get_dummies(train_data, columns=cat_cols)
test_data = pd.get_dummies(test_data, columns=cat_cols)

# تطبیق ستون‌های آموزش و آزمون
test_data = test_data.reindex(columns=train_data.drop(columns='TripReason').columns, fill_value=0)
```

```python
train_data['CancelTime'].fillna(pd.Timestamp('2100-01-01'), inplace=True)
test_data['CancelTime'].fillna(pd.Timestamp('2100-01-01'), inplace=True)
```

```python
bool_cols = ['Male', 'VehicleClass', 'Cancel']

# جایگذاری با False یا 0 (پیش‌فرض محافظه‌کارانه)
train_data[bool_cols] = train_data[bool_cols].fillna(False)
test_data[bool_cols] = test_data[bool_cols].fillna(False)
```

```python
numeric_cols = ['Price', 'CouponDiscount']

# جایگذاری با میانگین یا صفر
for col in numeric_cols:
    mean_val = train_data[col].mean()
    train_data[col].fillna(mean_val, inplace=True)
    test_data[col].fillna(mean_val, inplace=True)
```
