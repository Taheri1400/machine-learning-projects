<h2 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazirmatn" color="#0099cc">
پیش‌پردازش و مهندسی ویژگی
</font>
</h2>

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
    در این سوال شما می‌توانید از هر تکنیک پیش‌پردازش/مهندسی ویژگی که در فصل‌های گذشته آموختید، استفاده کنید.
    <br>
    تکنیک‌هایی که استفاده می‌کنید به شکل مستقیم مورد ارزیابی توسط سامانه داوری قرار <b>نمی‌گیرند.</b> بلکه همه آن‌ها در دقت مدل شما تاثیر خواهند گذاشت؛ بنابراین هر چه پیش‌پردازش/مهندسی ویژگی بهتری انجام دهید تا دقت مدل بهبود پیدا کند، امتیاز بیشتری از این سوال کسب خواهید کرد.
</font>
</p>

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from scipy.sparse import hstack

# خواندن داده‌ها
train = pd.read_csv('train_data.csv')
test = pd.read_csv('test_data.csv')

# --- 1. مدیریت مقادیر گمشده ---
# فرض می‌کنیم ستون‌های عددی مقادیر گمشده با 0 جایگزین شوند
num_cols = ['follower_count', 'following_count', 'is_business', 'is_verified', 'is_private']
train[num_cols] = train[num_cols].fillna(0)
test[num_cols] = test[num_cols].fillna(0)

# ستون‌های متنی را هم اگر گمشده بودند به رشته خالی تبدیل کنیم
text_cols = ['fullname', 'username', 'biography']
train[text_cols] = train[text_cols].fillna('')
test[text_cols] = test[text_cols].fillna('')

# --- 2. تبدیل متغیر هدف (gender) به عددی ---
le_gender = LabelEncoder()
train['gender_enc'] = le_gender.fit_transform(train['gender'])  # man=1, woman=0 مثلاً

# --- 3. کدگذاری ستون age به صورت Ordinal (چون دسته سنی هست و معنی عدد داره) ---
# در دیتاست age عددهای 1 تا 4 هستند و ترتیب دارند پس می‌توان به صورت عددی نگه داشت
# اگر خواستید one-hot کنید:
# ohe = OneHotEncoder(sparse=False)
# age_ohe = ohe.fit_transform(train[['age']])
# test_age_ohe = ohe.transform(test[['age']])
# ولی فعلاً اینجا از عددی استفاده می‌کنیم
train['age'] = train['age'].astype(int)
test['age'] = test['age'].astype(int)

# --- 4. استخراج ویژگی از متون -- TF-IDF روی biography و fullname ---
tfidf_bio = TfidfVectorizer(max_features=300, ngram_range=(1,2), stop_words='english')
tfidf_fullname = TfidfVectorizer(max_features=100, ngram_range=(1,1))

bio_train_tfidf = tfidf_bio.fit_transform(train['biography'])
bio_test_tfidf = tfidf_bio.transform(test['biography'])

fullname_train_tfidf = tfidf_fullname.fit_transform(train['fullname'])
fullname_test_tfidf = tfidf_fullname.transform(test['fullname'])

# --- 5. مهندسی ویژگی طول رشته متنی ---
train['fullname_len'] = train['fullname'].apply(len)
test['fullname_len'] = test['fullname'].apply(len)

train['biography_len'] = train['biography'].apply(len)
test['biography_len'] = test['biography'].apply(len)

# --- 6. آماده‌سازی ویژگی‌های عددی برای مقیاس‌بندی ---
numeric_features = ['age', 'follower_count', 'following_count', 'fullname_len', 'biography_len',
                    'is_business', 'is_verified', 'is_private']

scaler = StandardScaler()
train_numeric_scaled = scaler.fit_transform(train[numeric_features])
test_numeric_scaled = scaler.transform(test[numeric_features])

# --- 7. ترکیب همه ویژگی‌ها ---
# چون برخی ویژگی‌ها sparse matrix هستند (TF-IDF) و برخی dense، باید ترکیب کنیم
from scipy.sparse import csr_matrix

train_features = hstack([
    csr_matrix(train_numeric_scaled),
    bio_train_tfidf,
    fullname_train_tfidf
])

test_features = hstack([
    csr_matrix(test_numeric_scaled),
    bio_test_tfidf,
    fullname_test_tfidf
])

# حالا train_features و test_features آماده مدل‌سازی هستند
# و هدف هم:
y_train = train['gender_enc']

# --- 8. نمونه جدا کردن داده اعتبارسنجی ---
X_train, X_val, y_train_split, y_val = train_test_split(
    train_features, y_train, test_size=0.2, random_state=42, stratify=y_train
)

print(f'ابعاد داده آموزشی: {X_train.shape}')
print(f'ابعاد داده اعتبارسنجی: {X_val.shape}')
```

مدل‌سازی

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
    حال که داده را پاکسازی کرده و احتمالا ویژگی‌هایی را به آن افزوده یا از آن حذف کرده‌اید، وقت آن است که مدلی آموزش دهید که بتواند متغیر هدف این مسئله را پیش‌بینی کند.
</font>
</p>

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# فرض بر این است که داده ها به دو دسته آموزش و اعتبارسنجی تقسیم شده اند
# X_train, X_val, y_train_split, y_val

# تعریف مدل
model = LogisticRegression(max_iter=500, random_state=42)

# آموزش مدل روی داده آموزشی
model.fit(X_train, y_train_split)

# پیش‌بینی روی داده اعتبارسنجی
y_val_pred = model.predict(X_val)

# محاسبه دقت
acc = accuracy_score(y_val, y_val_pred)
print(f'Accuracy on validation set: {acc:.4f}')

# گزارش طبقه‌بندی
print(classification_report(y_val, y_val_pred, target_names=le_gender.classes_))

# ماتریس درهم ریختگی
print('Confusion Matrix:')
print(confusion_matrix(y_val, y_val_pred))

```

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">

    معیاری که برای ارزیابی عملکرد مدل انتخاب کرده‌ایم، <code>f1_score</code> نام دارد.
    <br>
    این معیار، سنجه ارزیابی کیفیت مدل شماست. به عبارت بهتر در سامانه داوری هم از همین معیار برای نمره‌دهی استفاده شده است.
    <br>
    پیشنهاد می‌شود با توجه به این معیار، عملکرد مدل خود را بر روی دادگان آموزش یا اعتبارسنجی ارزیابی کنید.

```python
from sklearn.metrics import f1_score

# پیش‌بینی روی داده اعتبارسنجی
y_val_pred = model.predict(X_val)

# دقت (Accuracy)
acc = accuracy_score(y_val, y_val_pred)
print(f'Accuracy on validation set: {acc:.4f}')

# F1-score برای هر کلاس (مقدار پیش‌فرض macro)
f1_macro = f1_score(y_val, y_val_pred, average='macro')
f1_weighted = f1_score(y_val, y_val_pred, average='weighted')
print(f'F1-score (macro): {f1_macro:.4f}')
print(f'F1-score (weighted): {f1_weighted:.4f}')

# گزارش طبقه‌بندی کامل شامل precision, recall, f1-score
print(classification_report(y_val, y_val_pred, target_names=le_gender.classes_))

```
